# 00 — Configuração do ambiente

## Objetivo
Preparar o ambiente do MVP no Databricks, utilizando o catálogo `ons_energia` e a arquitetura Medallion. Este notebook cria os schemas `bronze`, `silver` e `gold`, provisiona o Volume de entrada e verifica seu acesso.

**Pré-requisito:** o catálogo `ons_energia` deve existir e o usuário deve possuir permissões para criar schemas e Volume. A criação do catálogo não é realizada neste notebook. Execute esta etapa antes dos demais notebooks.

## 1. Preparação dos schemas

A instrução `CREATE SCHEMA IF NOT EXISTS` permite executar novamente a configuração sem falhar quando os schemas já existem. A consulta final lista os schemas disponíveis no catálogo.

```python

CATALOG = "ons_energia"

spark.sql(f"CREATE SCHEMA IF NOT EXISTS {CATALOG}.bronze")
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {CATALOG}.silver")
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {CATALOG}.gold")

spark.sql(f"SHOW SCHEMAS IN {CATALOG}").show()
```

## 2. Criação do Volume de entrada

O Volume `ons_energia.bronze.arquivos_ons` é o local de armazenamento do arquivo bruto. A criação condicional não substitui arquivos existentes. Os dados do Volume permanecem no Databricks e não são versionados automaticamente no GitHub.

```python

spark.sql(f"""
CREATE VOLUME IF NOT EXISTS {CATALOG}.bronze.arquivos_ons
COMMENT 'Armazenamento dos arquivos originais do ONS'
""")
```

A listagem confirma o acesso ao Volume. Antes de executar `01_ingestao_bronze`, disponibilize nele o arquivo `INTERRUPCAO_CARGA.csv`, mantendo exatamente esse nome. O dicionário da fonte pode ser mantido junto ao arquivo para referência documental.
